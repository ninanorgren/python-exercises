# Reverse complement DNA

```python
data = 'ATGGTCGAATGCGGTAACCGTAAATTGCGATCGCGCTAGGCCTAAACGTT'

code_dictionary = {'A': 'T',
                   'T': 'A',
                   'G': 'C',
                   'C': 'G'} 

reverse_sequence = data[::-1]
reverse_complement_list = []

for base in reverse_sequence:
    complement = code_dictionary[base]
    reverse_complement_list.append(complement)

reverse_complement_sequence = ''.join(reverse_complement_list)
reverse_complement_sequence
```

# Calculate GC content

```python
seq_a = 'ACGGCTGCCGGTGGCGGCCCGCGAATGCTTGCGC'
seq_b = 'TTATGTCGCTTAGGATTGCTCGTCGCGTCGGAGA'
seq_c = 'TTAGACAGCACGCTTGTGCGTAACGGACTCGCTA'
seq_d = 'AAGCGGCTGTGCCGCGTGGTGCGATGCCGTCGCC'

seq_dictionary = {'Sequence A': seq_a,
                  'Sequence B': seq_b,
                  'Sequence C': seq_c,
                  'Sequence D': seq_d}

for sequence in seq_dictionary:
    gc_count = 0
    for base in seq_dictionary[sequence]:
        if base == 'G' or base == 'C':
            gc_count += 1
    gc_content = gc_count / len(seq_dictionary[sequence])
    print(f"{sequence} has GC content: {gc_content}")
```

# Find motifs in DNA

```python
data = 'CTGAGCCTGGACGCCTGAGCTACGCCTTGCGAAACGCGTTCGCCTGGCTAACCGCGCCTA'
query_string = 'GCCT'

ix = 0
query_position_result = []

while ix < len(data):
    if data[ix:].startswith(query_string):
        query_position_result.append(ix)
    ix += 1
    
query_position_result
```
