# Create function for calculating heterozygosity

```python
frequencies = {'0/0': 4, '0/1': 3, '1/1': 2}

def calculate_heterozygosity(frequencies):
    heterozygots = frequencies['0/1']
    total = sum(frequencies.values())
    return heterozygots / total
    
heterogyzosity_rate = calculate_heterozygosity(frequencies)
heterogyzosity_rate
```

# Create a function that parses a line in a VCF file

```python
line = 'chr1    346     .       G       A       1451.38 PASS    AC=13;AF=0.325;AN=40   GT:AD:DP       0/0:15,1:16      0/0:16,0:16 0/1:15,12:28  0/1:15,5:20   0/1:18,5:23    1/1:3,21:24'

def parse_allele_frequencies(line):
    frequencies = {'0/0': 0, '0/1': 0, '1/1': 0}
    entries = line.strip().split()
    for genotype in entries[9:]:
        alleles = genotype.split(':')[0]
        if alleles in frequencies:
            frequencies[alleles] += 1
        else:
            print(f"Found illegal character {alleles}, skipping")
    return frequencies
    
frequencies = parse_allele_frequencies(line)
frequencies
```

# Create a function to read a VCF file

```python
path_to_vcf = '/mnt/c/Users/Nina/Documents/Projects/L_Andersson/mydata.vcf'
chromosome = 'chr1'
position = '167791'

def extract_snp_from_vcf(vcf, chrom, pos):
    vcf_file = open(vcf, 'r')
    for line in vcf_file:
        if not line.startswith('#'):
            entries = line.strip().split()
            if entries[0] == chrom and entries[1] == pos:
                return line
    
line = extract_snp_from_vcf(path_to_vcf, chromosome, position)
line
```
